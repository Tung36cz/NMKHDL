```python
# ==========================================
# 1. IMPORT THƯ VIỆN
# ==========================================

import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
```

```python
# ==========================================
# 2. ĐỌC DỮ LIỆU
# ==========================================

# Đọc file CSV
df = pd.read_csv("updated_pollution_dataset.csv")

# Hiển thị 5 dòng đầu
print("5 dòng đầu của dữ liệu:")
print(df.head())
```

```python
# ==========================================
# 3. KIỂM TRA THÔNG TIN DỮ LIỆU
# ==========================================

# Hiển thị thông tin dataset
print("\nThông tin dữ liệu:")
print(df.info())
```

```python
# ==========================================
# 4. KIỂM TRA GIÁ TRỊ THIẾU
# ==========================================

print("\nSố lượng giá trị thiếu:")
print(df.isnull().sum())
```

```python
# ==========================================
# 5. LABEL ENCODING
# Chuyển Air Quality từ text sang số
# ==========================================

# Tạo bộ mã hóa
le = LabelEncoder()

# Chuyển dữ liệu text sang số
df['Air Quality'] = le.fit_transform(df['Air Quality'])

# Hiển thị kết quả
print("\nAir Quality sau Encoding:")
print(df['Air Quality'].head())

# ==========================================
# GIẢI THÍCH
# Good -> 0
# Moderate -> 1
# Poor -> 2
# Hazardous -> 3
# ==========================================
```

```python
# ==========================================
# 6. TẠO FEATURE MỚI
# Tạo tỷ lệ PM2.5 / PM10
# ==========================================

# +1 để tránh chia cho 0
df['PM_ratio'] = df['PM2.5'] / (df['PM10'] + 1)

# Hiển thị kết quả
print("\nFeature mới PM_ratio:")
print(df[['PM2.5', 'PM10', 'PM_ratio']].head())
```

```python
# ==========================================
# 7. TÁCH DỮ LIỆU X VÀ y
# ==========================================

# X là dữ liệu đầu vào
X = df.drop('Air Quality', axis=1)

# y là nhãn cần dự đoán
y = df['Air Quality']

# Hiển thị kích thước dữ liệu
print("\nKích thước X:", X.shape)
print("Kích thước y:", y.shape)
```

```python
# ==========================================
# 8. CHUẨN HÓA DỮ LIỆU
# ==========================================

# Tạo bộ chuẩn hóa
scaler = StandardScaler()

# Chuẩn hóa dữ liệu
X_scaled = scaler.fit_transform(X)

# Hiển thị dữ liệu sau chuẩn hóa
print("\n5 dòng đầu sau chuẩn hóa:")
print(X_scaled[:5])
```

```python
# ==========================================
# 9. KẾT LUẬN
# ==========================================

print("\nHoàn thành Feature Engineering")

# Đã thực hiện Feature Engineering gồm:

#  -Mã hóa biến mục tiêu bằng Label Encoding
#  -Chuẩn hóa dữ liệu bằng StandardScaler
#  -Tạo đặc trưng mới PM_ratio từ PM2.5 và PM10

# Dữ liệu sau xử lý đã sẵn sàng cho các mô hình Machine Learning.
# ==========================================
```
